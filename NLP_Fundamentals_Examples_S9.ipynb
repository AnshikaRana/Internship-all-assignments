## Text Cleaning

Text cleaning is the process of **removing noise from raw text** so that machines can process it effectively.

### Why is text cleaning important?
Raw text often contains:
- Uppercase and lowercase variations
- Punctuation and symbols
- Numbers and emojis
- Extra white spaces

These add **noise**, not meaning.

### Common text cleaning operations:
- Lowercasing text
- Removing punctuation
- Removing numbers
- Removing special characters
- Normalizing white spaces

Clean text leads to:
- Smaller vocabulary
- Better model performance
- Faster processing

## Tokenization

Tokenization is the process of **breaking text into smaller units** called tokens.

### Types of Tokenization:
1. **Word Tokenization** – splits text into words
2. **Sentence Tokenization** – splits text into sentences

### Why tokenization is important:
- Models do not understand raw text
- Tokens are the basic units used in NLP
- Most NLP operations work on tokens, not full text

Example:
"I love NLP" → ["I", "love", "NLP"]


## 4. Word Tokenization

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')             # Download tokenizer model

text1 = "NLP is amazing"
print(word_tokenize(text1))        # Split sentence into words

text2 = "I love AI. Added more content"
print(word_tokenize(text2))
```

## 5. Stopword Removal

Stopwords are **very common words** that usually do not add much meaning.

Examples:
- is, the, and, of, to, in

### Why remove stopwords?
- They appear in almost every document
- They increase vector size
- They dilute important information

After stopword removal:
- Text becomes more meaningful
- Feature space reduces


```python
from nltk.corpus import stopwords

nltk.download('stopwords')          # Download stopword list

stop_words = set(stopwords.words('english'))  # Load English stopwords

tokens1 = word_tokenize("NLP is very interesting")
print(tokens1)
print([w for w in tokens1 if w not in stop_words])
# Removes common words like 'is', 'very'

tokens2 = word_tokenize("AI is the future not")
print(tokens2)
print([w for w in tokens2 if w not in stop_words])
```

## 6. Stemming

Stemming reduces words to their **root form** using rule-based methods.

Examples:
- running → run
- studies → studi
- playing → play

### Characteristics of stemming:
- Fast
- Rule-based
- Output may not be a real dictionary word

### Use cases:
- Search engines
- Information retrieval


```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()           # Initialize stemmer

words1 = ['running', 'jumps']
print([stemmer.stem(w) for w in words1])  # Reduce to root form

words2 = ['studies', 'playing']
print([stemmer.stem(w) for w in words2])
```

## 7. Lemmatization

Lemmatization reduces words to their **dictionary base form (lemma)**.

Examples:
- running → run
- studies → study
- better → good

### Characteristics of lemmatization:
- Uses vocabulary and grammar
- More accurate than stemming
- Slower than stemming

### Preferred in:
- Chatbots
- NLP pipelines
- Semantic analysis


```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')             # Dictionary for lemmatization

lemmatizer = WordNetLemmatizer()     # Initialize lemmatizer

words1 = ['running', 'cars']
print([lemmatizer.lemmatize(w) for w in words1])

words2 = ['studies', 'better']
print([lemmatizer.lemmatize(w) for w in words2])
```

## 8. One-Hot Encoding

One-Hot Encoding represents each word as a **binary vector**.

### How it works:
- Vocabulary size = vector length
- Each word gets a unique index
- Presence → 1
- Absence → 0

Example Vocabulary:
["ai", "love", "nlp"]

"nlp" → [0, 0, 1]
"love" -> [0, 1,0]

### Limitations:
- Very large vectors
- No semantic meaning
- Sparse representation


```python
import numpy as np

vocab = ["ai", "data", "learning", "nlp"]

def one_hot(word, vocab):
    vector = np.zeros(len(vocab))
    if word in vocab:
        vector[vocab.index(word)] = 1
    return vector

for word in vocab:
    print(word, "→", one_hot(word, vocab))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = ["I love NLP love", "I love AI"]

vectorizer = CountVectorizer(binary=True)  # Binary = presence/absence
X = vectorizer.fit_transform(sentences)    # Learn vocab and encode

print(vectorizer.get_feature_names_out())  # Vocabulary
print(X.toarray())                         # One-hot vectors
```

### ⚠️ Important: Core Difference Between the Two Outputs

Although both outputs look like *one-hot encoding*, they represent **different things**.

- **Word-level One-Hot Encoding**  
  - Each **individual word** is represented by its own vector  
  - Vector has **exactly one `1`** and rest `0`s  
  - Used to uniquely identify a word  
  - Example:  
    `nlp → [0, 0, 0, 1]`

- **Document-level Binary Encoding (CountVectorizer)**  
  - Each **document (sentence)** is represented by a vector  
  - Vector shows **presence (1) or absence (0)** of words  
  - A row can have **multiple `1`s** because a document has many words  
  - This is actually a **binary Bag of Words**, not true one-hot encoding

**In short:**  
Word-level one-hot encodes **words**, while CountVectorizer with `binary=True` encodes **documents**.

## 9. Bag of Words

Bag of Words represents text by **counting word frequency**.

### Key ideas:
- Ignores grammar and word order
- Considers how often a word appears
- Creates a document-word matrix

Example:
"I love love NLP" → love = 2

### Advantages:
- Simple
- Easy to implement

### Limitations:
- No meaning
- Common words dominate


```python
vectorizer = CountVectorizer()         # Counts word frequency

sentences = ["I love love NLP. I also love pizza.", "I love AI"]
X = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# Import the required class
from sklearn.feature_extraction.text import CountVectorizer

# Input documents
documents = [
    "I love love NLP",
    "I love AI",
    "NLP and AI are powerful"
]

# Initialize CountVectorizer with stopword removal
vectorizer = CountVectorizer(stop_words='english')

# Fit the vectorizer and transform documents into BoW matrix
bow_matrix = vectorizer.fit_transform(documents)

# Display the learned vocabulary
print("Vocabulary (column order):")
print(vectorizer.get_feature_names_out())

# Display the Bag of Words matrix
print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

## 10. TF-IDF

## TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF measures **how important a word is** to a document relative to a collection of documents.

### Intuition:
- Words frequent in one document but rare overall are important
- Common words are down-weighted

### Components:
- TF (Term Frequency): how often a word appears in a document
- IDF (Inverse Document Frequency): how rare the word is across documents

### Why TF-IDF is better than BoW:
- Reduces impact of common words
- Highlights meaningful terms


```python
# Import TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Input documents (same as previous examples)
documents = [
    "I love love NLP",
    "I love AI",
    "NLP and AI are powerful"
]

# Initialize TF-IDF vectorizer with stopword removal
tfidf_vectorizer = TfidfVectorizer(stop_words='english')

# Fit the model and transform documents into TF-IDF matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Display the learned vocabulary (column order)
print("Vocabulary (column order):")
print(tfidf_vectorizer.get_feature_names_out())

# Display the TF-IDF matrix
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()            # Initializes TF-IDF
sentences = ["I love NLP", "I love AI"]

X = tfidf.fit_transform(sentences)
print(tfidf.get_feature_names_out())
print(X.toarray())                   # Importance-based scores
```

#One-Hot Encoding vs Bag of Words

```python
# Import CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

# Input documents (note: 'love' appears twice in Document 1)
documents = [
    "I love love NLP",
    "I love AI",
    "NLP and AI are powerful"
]

# ---------- Bag of Words ----------
bow_vectorizer = CountVectorizer(stop_words='english')
bow_matrix = bow_vectorizer.fit_transform(documents)

print("Bag of Words Vocabulary:")
print(bow_vectorizer.get_feature_names_out())
print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# ---------- One-Hot Encoding ----------
one_hot_vectorizer = CountVectorizer(stop_words='english', binary=True)
one_hot_matrix = one_hot_vectorizer.fit_transform(documents)

print("\nOne-Hot Vocabulary:")
print(one_hot_vectorizer.get_feature_names_out())
print("\nOne-Hot Matrix:")
print(one_hot_matrix.toarray())
```

## 13. Word2Vec

Word2Vec is a technique that represents words as **dense numerical vectors** such that
**words with similar meanings have similar vectors**.

Unlike One-Hot, Bag of Words, or TF-IDF:
- Vectors are **dense**, not sparse
- Vectors **capture semantic meaning**
- Similar words are **closer in vector space**

Example intuition:
- "king" and "queen" → similar vectors  
- "AI" and "NLP" → closer than "AI" and "banana"

Word2Vec learns these vectors by analyzing **context words**.


### How Word2Vec Learns Meaning

Word2Vec is based on the idea:
> "You shall know a word by the company it keeps."

There are two main architectures:
1. **CBOW (Continuous Bag of Words)**  
   - Predicts a word from its surrounding context

2. **Skip-gram**  
   - Predicts surrounding context words from a target word

In practice, Skip-gram works better for small datasets.


```python
# Install gensim (if not already installed)
!pip install -q gensim
```

```python
# Import Word2Vec
from gensim.models import Word2Vec

# Small example corpus (list of tokenized sentences)
sentences = [
    ["i", "love", "nlp"],
    ["i", "love", "ai"],
    ["nlp", "and", "ai", "are", "related"],
    ["machine", "learning", "uses", "ai"],
    ["deep", "learning", "is", "part", "of", "ai"]
]

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=5,     # Size of word vectors
    window=2,          # Context window
    min_count=1,       # Include all words
    sg=1               # Skip-gram model
)
```

```python
# Get vector for a word
print("Vector for 'ai':\n", model.wv["ai"])
print("\nVector for 'nlp':\n", model.wv["nlp"])
```

```python
# Find words similar to 'ai'
model.wv.most_similar("ai")
```
